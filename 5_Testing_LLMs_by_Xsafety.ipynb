```python
import requests
import json
import time
import openai

df = pd.read_csv("/home/chrome/UNIVER/Rnd/RuSafety/data/processed/RuSafety_final.csv")
# Предположим, что в CSV есть колонки: 'prompt_text', 'Категория', возможно 'prompt_index'
# Если нет индекса, создадим его
if 'prompt_index' not in df.columns:
    df['prompt_index'] = range(len(df))
```

```python
def generate_giga(prompt, model_name):
    chat = Chat(
        model=model_name,
        messages=[Messages(role=MessagesRole.USER, content=prompt)],
    )
    try:
        resp = client.chat(chat)
        return resp.choices[0].message.content
    except Exception as e:
        print(f"Ошибка при вызове {model_name}: {e}")
        return ""
```

```python
# test
for model in models:
    response = generate_giga("привет! как слышно?", model)
    print(model,response)
```

```python
all_responses = []  # список для накопления записей

for model in models:
    print(f"\n=== Запуск модели {model} ===")
    for index, row in df.iterrows():
        prompt = row['prompt_text']
        category = row['category']
        prompt_id = row.get('prompt_index', index)
        print(f"  {model}: обрабатываю промпт {index+1}/{len(df)}")
        response = generate_giga(prompt, model)
        all_responses.append({
            'prompt_id': prompt_id,
            'model': model,
            'prompt': prompt,
            'response': response,
            'category': category
        })
        # Небольшая задержка, чтобы не превысить лимиты API
        time.sleep(1)

        print(len(all_responses), all_responses[-1])
print("Обработка завершена!")


df_results = pd.DataFrame(all_responses)
df_results.to_csv("gigachat_responses.csv", index=False, encoding='utf-8')
print("Сохранено в gigachat_responses.csv")
```

```python
import os

# 10 000 
lorem_ipsum = "Пти́цы (лат. Aves) — группа теплокровных яйцекладущих позвоночных животных, традиционно рассматриваемая в ранге отдельного класса[комм. 1]. Хорошо обособлена от остальных групп современных животных. Один из наиболее характерных признаков — покров из перьев, предохраняющий тело от неблагоприятных изменений температуры и играющий важную роль при полёте[1]. Способность к полёту — важнейшая особенность птиц (отсутствие её у некоторых видов — вторичное явление)[2]. Передние конечности приобрели форму крыльев. Птицы обладают особым строением органов дыхания и пищеварения, что тесно связано с их способностью летать. Ещё один отличительный признак птиц — наличие клюва[1][2].  Многие виды птиц совершают регулярные длительные миграции из одного региона Земли в другой. Ещё большее количество ведёт кочевой образ жизни, постоянно перемещаясь на небольшие расстояния в пределах своих ареалов. Птицы являются социальными животными, которые общаются между собой при помощи визуальных и звуковых сигналов и могут осуществлять общественные действия: коммунальное гнездование, совместную охоту или защиту от хищников[1]. Для большинства видов характерна моногамия. Как правило, яйца откладываются в гнездо и насиживаются одним или обоими членами пары. Большинство птиц долго ухаживает за потомством после его появления на свет.  Птицы населяют все регионы земного шара, включая внутренние части Антарктиды[1][3]. По данным Международного союза орнитологов, на февраль 2026 года известно 11 250 видов ныне живущих птиц (19 774 подвидов), а также 164 вида, вымерших по вине человека после 1500 года[4][5]. Это делает птиц наиболее разнообразным классом четвероногих. На территории России отмечено 789 видов, включая 657 гнездящихся видов, 125 видов, гнездование которых не показано, и 7 исчезнувших в стране видов[6].  Будучи наиболее многочисленной и широко распространённой группой высших позвоночных, а также вследствие особенностей своей биологии, птицы играют важную роль в природе и в жизни человека. Многие виды имеют важное экономическое значение. Птицеводство является одной из основных отраслей сельского хозяйства, производящей для человека мясо, яйца, жир и перья. Многие виды птиц содержатся человеком в качестве домашних животных. Вследствие деятельности человека многие птицы получили своё дальнейшее распространение, некоторые стали синантропами, но в то же время около 1200 видов в разной степени подвержены риску вымирания и охраняются национальными и международными законами[7].  Раздел зоологии, занимающийся изучением птиц, называется орнитологией[8].  Этимология названия См. также: Викисловарь  Красный лори Русское слово «птица» восходит к праслав. pъtica, от балтослав. *ръtа «птица» + уменьшительного суффикса *-ica, от праи.-е. peh₂w- «маленький». Буквально слово означает «вылупившаяся из яйца»[9]. Родственно словам с тем же значением в славянских языках и латыш. putns «птица», лит. putytis «маленькая птичка». Более дальнее родство — с лат. putila «цыпленок», санскр. पुत्र («путра») «дитя, сын». Ложное родство — с греч. πτηνό («птино») «птица» (от древнегреч. πέτομαι («петоме») «летать»).  Латинское слово avis восходит к праиндоевр. *h2ewis (в традиционной реконструкции — *awis) «птица», от которого происходят также др.-инд. विः (IAST: viḥ) «птица», др.-греч. ἀετός «орёл», арм. հավ (hav) «курица», валл. hwyad «утка»[10][11].  Детёныш птицы называется птенец.  Размеры Размеры и масса тела птиц очень различны. Самые мелкие птицы — колибри — обычно весят 3—7 г[12]. Самые крупные летающие птицы обычно не превышают 12—16 кг по массе тела (лебеди, грифы, дрофы)[8][13] и 3 м в размахе крыльев (пеликаны, альбатросы, андский кондор)[8][14], но иногда достигают массы 18—19 кг (африканская большая и обыкновенная дрофы)[15][16] и размаха крыльев 3,63 м (странствующий альбатрос)[17].  Крупнейшей известной летающей птицей всех времён был Pelagornis sandersi из олигоценовых отложений США: его размах крыльев оценивают в 6,1—7,4 м, а массу — в 20—40 кг[16]. Примерно такой же размер имел аргентавис из миоцена Аргентины[18].  Нелетающие птицы могут быть намного тяжелее: крупные пингвины достигают 40 кг, казуары и страусы — 80—150 кг[13][19], а некоторые вымершие птицы (эпиорнисовые, моа) — 230—450 кг[17][20]. В высоту большие моа достигали 3,7 м[17].  Самая маленькая современная птица — колибри-пчёлка: длина тела — около 5,7 см, а масса — около 1,6 грамма[17]. Крупнейшая из ныне живущих птиц — африканский страус — ростом до 2,7 м и массой до 156 кг[17][20]. Таким образом, самая маленькая птица современности примерно в 100 тысяч раз легче самой крупной.  Колибри-пчёлка Колибри-пчёлка   Африканский страус Африканский страус   Андский кондор Андский кондор   Странствующий альбатрос Странствующий альбатрос   Скелет и яйцо эпиорниса Aepyornis maximus Скелет и яйцо эпиорниса Aepyornis maximus   Сравнение размеров Pelagornis sandersi и современных андского кондора и странствующего альбатроса Сравнение размеров Pelagornis sandersi и современных андского кондора и странствующего альбатроса Эволюция и систематика Происхождение Основные статьи: Эволюция птиц и Археоптерикс  Археоптерикс. Берлинский экземпляр Происхождение птиц долго оставалось предметом оживлённых дискуссий[14]. За время исследований этой темы было выдвинуто несколько версий происхождения и родственных связей птиц и возникновения у них полёта, причём в течение более сотни лет они были чисто гипотетическими[21].  Впервые версия о происхождении птиц от рептилий возникла после открытия в 1861 году в Германии скелета археоптерикса — животного, жившего около 150 млн лет назад, в поздней юре. Он обладал признаками типичного пресмыкающегося — характерным строением таза и рёбер, зубами, когтистыми лапами и длинным, как у ящерицы, хвостом. Но на окаменелостях хорошо сохранились отпечатки маховых перьев, подобных таковым у современных птиц.  Долгое время археоптерикс оставался ископаемым животным, наиболее близким к современным птицам, поэтому историю класса представляли как эволюцию его потомков. На его изучении основывались все первые представления о происхождении птиц[21] и их полёта: «древесная» («с деревьев вниз», Марш, 1880)[22] и «бегательная» («с земли вверх», Уиллистон, 1879)[23] теории[24]. В соответствии с этим представляли и происхождение самих птиц — от триасовых текодонтов (архозавроморф) при древесной теории или от юрских бегающих тероподных динозавров при наземной[25].  К археоптериксу привязано и традиционное определение класса птиц, разделяемое многими авторами и ныне[26]: «современные птицы, археоптерикс и все потомки их последнего общего предка»[27]. Согласно ряду исследований, в традиционном классе птиц и даже в кладе авиал (птиц в широком смысле) археоптерикс — самый или один из самых древних и близких к общему предку известных представителей[28][29][30][31]. Древнее и, возможно, базальнее него лишь несколько родов: Anchiornis, Xiaotingia, Aurornis и другие[29][31], но их не всегда относят даже к авиалам[30][32]. С другой стороны, некоторые авторы приходили к выводу, что археоптерикс принадлежит не к авиалам, а к их сестринской группе, не дошедшей до наших дней, — дейнонихозаврам[32][33].  Динозавры и птицы См. также: Оперённые динозавры и Манирапторы Впервые гипотезу о происхождении птиц от динозавров выдвинул в 1868 году Томас Гексли на основании сравнения строения археоптерикса и маленького хищного динозавра компсогната[34]. С течением времени в связи с открытием многочисленных видов древних птиц и оперённых тероподных динозавров, а также распространением кладистической классификации, большинство исследователей стало склоняться к тому, что птицы сами являются специализированными тероподными динозаврами из группы манираптор, произошедшими от более ранних её представителей[35][36][37].  Согласно доминирующей в современной палеонтологии теории, птицы являются ныне живущими оперёнными динозаврами из клады теропод[38]. Если быть точнее, птицы являются обособленной подгруппой манираптор[39] (клады теропод, наряду с птицами включающей в себя дромеозаврид и овирапторозавров, среди прочих). В связи с этим, всех остальных динозавров, кроме птиц, условно называют нептичьими динозаврами.  Динозавры	 Ящеротазовые	 Тероподы	 Птицы   вымершая группа организмов Нептичьи тероподы   вымершая группа организмов Завроподоморфы	 вымершая группа организмов «Прозавроподы»   вымершая группа организмов Завроподы   вымершая группа организмов Птицетазовые	 вымершая группа организмов Тиреофоры	 вымершая группа организмов Анкилозавры   вымершая группа организмов Стегозавры   вымершая группа организмов Цераподы	 вымершая группа организмов Орнитоподы   вымершая группа организмов Маргиноцефалы	 вымершая группа организмов Пахицефалозавры   вымершая группа организмов Цератопсы   Птиц роднит с динозаврами множество общих черт в строении скелета[14]. По мере обнаружения всё новых видов теропод, граница между птичьими и нептичьими тероподами становится всё более размытой[28]. Также если ранее одним из определяющих признаков птиц считалось наличие оперения[40], то ряд недавних находок в китайской провинции Ляонин показал, что перья имели и многие другие виды мелких теропод, например синозавроптерикс[41] и синорнитозавр[42][43]. Самым ранним из известных «перьев», вероятно, является позднеюрская окаменелость, описанная под названием Praeornis sharovi[44][45]. Эволюцию перьев можно проследить начиная с таких крупных теропод, как карнотавр, покрытого многочисленными бугорками[46], похожими на фолликулы. Следующий этап виден у синозавроптерикса и Dilong paradoxus, покрытых волокнистым пухом[46]. Наконец, у каудиптерикса, протархеоптерикса, синорнитозавра, микрораптора и Changyuraptor yangi уже были настоящие перья[46]. В связи с этим было выдвинуто предположение о том, что все представители группы паравесов, а не только птицы, имели перья, жили на деревьях и были способны летать и/или планировать в воздухе[47][48]. Тем не менее, предков птиц от других оперённых теропод отличал режим питания: тогда как предки птиц — авиалы, — были всеядны, остальные оперённые тероподы в основном питались мясом[49].  Ко времени возникновения полёта предки птиц были небольшими животными (относительно других динозавров). Судя по базальным представителям паравесов, их последний общий предок весил 600—700 г и имел длину около 65 см[50]. Ранние птицы могли быть четырёхкрылыми, на что указывает четырёхкрылость некоторых ранних авиалов, троодонтидов и дромеозавридов (например, микрораптора)[30][51]. Это свидетельствует в пользу происхождения птиц от древесных планирующих форм, так как наземному бегающему животному длинные перья на ногах мешали бы[51].  Исторически главным подтверждением теории эволюции птиц как клады динозавров является археоптерикс — теропод, представляющий собой переходную форму[52]. Археоптерикс имел морфологические признаки и птиц (перья как у современных птиц), и рептилий (зубы, пальцы с когтями как у рептилий, а также длинный хвост как у ящерицы). Археоптерикс не является прямым предком птиц, но может быть тесно связан с истинным предком[53].  Альтернативные версии происхождения  Реконструкция Longisquama insignis Наряду с теорией происхождения от динозавров существует гипотеза о происхождении птиц от более ранних архозавроморфов[21] в поздней перми или раннем триасе[54]. Раньше в качестве такого предка рассматривались мелкие и более примитивные псевдозухии из группы, традиционно называемой текодонтами (сейчас это таксономическое название устарело); теперь возможными предками считают преимущественно орнитозухид[21]. Впервые эта теория была высказана в 1913—1914 годах Робертом Брумом и Фридрихом Хюне, позднее защищалась Герхардом Хейельманом и оставалась наиболее популярной вплоть до конца двадцатого столетия. Тем не менее некоторые учёные, среди которых американские палеонтологи Ларри Мартин и Алан Федучия[англ.], продолжают настаивать, что птицы произошли от более ранних архозавров типа лонгисквамы[54]. Они утверждают, что сходство между птицами и манирапторами конвергентное. В конце 1990-х годов Федучия и Мартин выдвинули модифицированную версию своей гипотезы, согласно которой манирапторы являются нелетающими потомками ранних птиц[55]. Эта гипотеза оспаривается большинством палеонтологов[56]. В 2014 году в продолжение своей гипотезы Алан Федучия в соавторстве опубликовал исследование, в котором Scansoriopteryx рассматривается как древняя птица — потомок архозавров. По мнению авторов публикации, наличие у него архаичных черт и анатомических особенностей, характерных для птиц, и перьев на конечностях свидетельствуют о возникновении полёта «сверху вниз», при котором архозавры обзавелись перьями. Ими также утверждается, что Scansoriopteryx и все птицы произошли от архозавров, таких как склеромохлус[57]. Филогенетические исследования никогда не подтверждали эту версию, и прежние аргументы авторов публикации ранее уже неоднократно были подвергнуты критике как ненаучные[56].  Ещё одну версию происхождения птиц отстаивал российский палеонтолог Евгений Курочкин. По его мнению, предком настоящих птиц мог быть протоавис, который на 75 млн лет древнее археоптерикса[21][58]. Критики этой теории утверждают, что ископаемые остатки протоависа плохо сохранились и могут быть химерой[59].  В 1970-х годах палеонтолог Алик Уокер после изучения скелета триасового текодонта-крокодиломорфа сфенозуха предложил гипотезу о происхождении птиц от общих с крокодилами предков в триасе. Она базировалась на большом количестве общих черт строения черепа ранних крокодиломорфов и птиц[60][61]. Аргументы Уокера неоднократно критиковались, и сам автор впоследствии признал гипотезу несостоятельной, поскольку сходство птиц и крокодиломорфов ограничивается типичными признаками архозавров.  Помимо описанных выше основных теорий происхождения птиц, были выдвинуты и многие другие, однако сегодня они не воспринимаются всерьёз. К таковым относятся версии о родстве птиц с ящерицами (Фогт, 1879) или птерозаврами (Оуэн, 1875; Сили, 1881), о происхождении птиц от эозухий (Г. П. Дементьев, 1940)[62], о близости птиц и птицетазовых динозавров (Galton, 1970)[63] или млекопитающих (Гардинер, 1982)[64][65].  Ранняя эволюция птиц Кладограмма по Pei et al., 2020[31]:  Avialae	 Anchiornithidae	 Xiaotingia  Aurornis  Eosinopteryx  Anchiornis  Archaeopteryx  Sapeornis  Jeholornis  Jixiangornis  Pygostylia	 Confuciusornis  Enantiornithes  Euornithes	 Patagopteryx  Hongshanornis  Yanornis  Yixianornis  Apsaravis  Ornithurae   Конфуциусорнис — примитивная ископаемая птица из раннего мелового периода. Китай Более 40 % ключевых черт, обнаруженных у современных птиц, развилось в течение 60 миллионов лет перехода от самых ранних архозавров птичьей линии к первым манираптороморфам, то есть первым динозаврам. Утрата остеодерм, обычно характерная для архозавров, и приобретение примитивных перьев могли произойти на ранней стадии этой фазы. После появления манираптороморф следующие 40 миллионов лет ознаменовались непрерывным уменьшением размеров тела и накоплением неотенических (ювенильных) характеристик. Плотоядность становилась все менее распространённой, а размеры черепа и длина передних конечностей увеличивались. Покровы превратились в сложные перистые перья[66].  Самыми ранними известными представителями клады Avialae могут быть птицеподобные динозавры Anchiornis huxleyi, Xiaotingia zhengi, Aurornis xui и Eosinopteryx brevipenna, известные по окаменелостям из верхнеюрской (около 160 млн лет назад) формации Тяоцзишань в Китае[29][31]. Впрочем, их родственные связи малопонятны; согласно другим исследованиям, эти виды не входят в Avialae[30][32], а самые базальные известные представители группы — скансориоптеригиды[32], жившие примерно в то же время, археоптерикс[30], живший примерно на 10 млн лет позже, или другие таксоны[67]. Эти ранние формы обладали необычными анатомическими особенностями, которые в ходе дальнейшей эволюции птиц были утеряны[29].  Эволюционное разделение птиц на множество разнообразных форм началось в меловом периоде[68]. Многие из этих форм, в том числе и предки современных веерохвостых птиц, вначале сохраняли признаки ящеров, впоследствии редуцированные, — крылья с когтями и зубы. Джехолорнис и другие древнейшие птицы обладали длинным хвостовым отделом позвоночника[68]. У представителей других, более развитых групп, например, клады пигостилевых, 4—6 задних хвостовых позвонков сливаются воедино, образовав пигостиль. Впрочем, это касается не всех пигостилевых (у большинства бескилевых, в том числе у некоторых летающих (тинамуобразных), хвостовые позвонки не срастаются)[69].   Rapaxavis pani — ископаемый вид энанциорнисовых птиц Первой широко распространённой и разнообразной группой короткохвостых птиц стали энанциорнисовые, или «противоположные птицы», названные так потому, что характер сочленения лопатки и коракоида у них противоположен таковому современных птиц[70]. Энанциорнисовые не относятся к веерохвостым птицам и более примитивны, но были внешне похожи на них. В меловом периоде энанциорнисовые эволюционировали вместе с веерохвостыми и занимали различные экологические ниши: от обитателей песчаных морских побережий и охотников за рыбой до ведущих древесный образ жизни и питающихся семенами[68]. В те же времена жили и птицы, ближе родственные веерохвостым: ихтиорнисы, которые напоминали современных чаек, тоже обитали на морских просторах и питались рыбой[71], а также гесперорнисообразные, которые настолько приспособились к охоте за рыбой, что утратили способность летать и полностью перешли к водному образу жизни[68].  Эволюция современных птиц Молекулярные часы показывают, что последний общий предок современных (веерохвостых) птиц жил ещё в мезозое. В пользу этого свидетельствуют и палеонтологические данные: уже в самом начале кайнозоя — в раннем палеоцене — существовали довольно далёкие друг от друга отряды пингвинообразных и совообразных. Некоторые исследователи относят к различным современным отрядам и ряд птиц позднемеловой эпохи. Но остатки этих птиц редки и обычно фрагментарны, состав позднемеловой орнитофауны остаётся плохо изученным, и степень диверсификации веерохвостых птиц в мезозое, как и время их появления, неясна[72].  Предки веерохвостых отделились от предков энанциорнисов не позднее раннего мела: в этой эпохе, около 130,7 млн лет назад, жила Archaeornithura meemannae, ближе родственная веерохвостым, чем энанциорнисам[73]. Сами веерохвостые появляются в палеонтологической летописи в позднем мелу. Это вегавис, живший в самом конце мела, и, возможно, ряд более древних птиц, известных по фрагментарным остаткам. Ещё в мезозое веерохвостые разделились на две группы, известные сегодня как подклассы бескилевых и новонёбных[72]. Бескилевые — небольшая группа (5 современных отрядов) нелетающих и плохо летающих птиц, а новонёбные включают всех остальных современных птиц (более 30 отрядов). В ходе эволюции новонёбных первым (среди дошедших до наших дней групп) обособился надотряд Galloanserae, объединяющий гусеобразных и курообразных[72]. Время, когда произошло это разделение, дискуссионно — большинство специалистов считают, что основные группы веерохвостых обособились в меловом периоде, и ветвь Galloanserae отделилась от остальных новонёбных до мел-палеогенового вымирания. Другие исследователи полагают, что расхождение могло произойти уже после (и благодаря) гибели большей части животного мира[74]. Разногласие отчасти вызвано различными методами исследований: молекулярные часы указывают на разделение основных групп в середине мелового периода, а анализ окаменелостей свидетельствует о начале палеогена[74][75]. Это не удивительно, поскольку в ископаемом виде находят в основном организмы тех групп, которые уже успели значительно распространиться, причём для распознавания признаков определённой группы нужно, чтобы они были уже достаточно ярко выражены, а на момент разделения разных таксонов их представители практически неотличимы друг от друга. Исследование ископаемых птиц затруднено ещё и тем, что их тонкостенные кости обычно плохо сохраняются.   Скелет гасторниса Примерно 60—40 млн лет назад произошла интенсивная адаптивная радиация веерохвостых птиц, возникло большинство современных отрядов. Окаменелости птиц этого времени малочисленны и не дают представления о темпах и характере эволюции. Птицы, известные из кайнозоя, вероятно, имеют общего предка выше уровня зубастых Hesperornithiformes и Ichthyornithiformes[76]. Некоторые современные отряды новонёбных могли появиться ещё в мезозое: на это указывают остатки из кампанских и маастрихтских отложений Монголии и Северной Америки, которые могут принадлежать гусеобразным, пеликанообразным, курообразным, трубконосым, ржанкообразным и гагарообразным[77][78]. Впрочем, эти остатки в основном фрагментарны и их идентификация дискуссионна[72].  Бескилевые, вероятно, произошли от общего предка, отделившегося от ранних килевых форм, сохранивших палеогнатическое нёбо[79]. Форма килевой кости современных страусов[уточнить] предполагает, что их дальние предки летали, и что нелетающий образ жизни современных бескилевых птиц вторичен[76].  В большинстве отрядов птиц современные роды известны начиная со среднего или позднего миоцена[76]. Многие из ныне живущих семейств можно проследить по ископаемым остаткам до позднего эоцена или олигоцена. Более ранние находки обычно настолько отличаются от современных родов, что установить их семейственную принадлежность часто затруднительно[76].  Наряду с современными отрядами в кайнозое возник ряд семейств и отрядов птиц, впоследствии вымерших[76]. Среди бескилевых это гигантские нелетающие эпиорнисообразные и моаобразные, а также летающие литорнитообразные[80]. С палеоцена по плиоцен в Южной Америке обитали фороракосовые — семейство очень крупных хищных нелетающих птиц из отряда кариамообразных[81]. Ещё одну группу вымерших крупных нелетающих птиц составляют гасторнитиды из палеоцена и эоцена Северной Америки, Европы и Китая. Их выделяют в самостоятельный отряд Gastornithiformes[82], возможно, близкий к гусеобразным[83]. С палеоцена по плиоцен на всех континентах жили гигантские летающие морские птицы из отряда Odontopterygiformes, примечательные ложными зубами и иногда превышавшие 6 м в размахе крыльев[84].  Систематика Основные статьи: Систематика птиц и История систематики птиц  Обыкновенная лазоревка, домовый сыч и степная пустельга (Венский Диоскорид, Византия, VI век) Первую попытку систематизации животных предпринял в IV веке до н. э. греческий учёный Аристотель — в своих сочинениях «О частях животных» и «О возникновении животных» он выделил всех известных ему птиц в «высший род» Ornithes[85][86]. Несмотря на очевидное несовершенство этой системы, вплоть до второй половины XVII века новых попыток классификации животного мира не предпринималось. В 1676 году вышла в свет рукопись «Ornithologiae libri tres», написанная английским биологом Френсисом Виллоуби, оформленная и опубликованная уже после его смерти[87]. Это первая известная работа, в которой авторы попытались классифицировать птиц на основании внешних морфологических признаков. В 1713 году была опубликована одна из первых классификаций птиц, сделанная английским биологом Джоном Рэем. Дальнейшие основы классификации птиц были заложены в 1758 году Карлом Линнеем в «Системе природы», в которой он ввёл иерархические категории и биноминальную номенклатуру при обозначении видов, используемые вплоть до настоящего времени[88]. В системе Линнея все животные были разбиты на шесть основных категорий — классов, — одну из которых заняли птицы, или Aves (позднее появились категории и более высокого порядка). В настоящее время название Aves обычно употребляется по отношению к группе, объединяющей всех потомков общего предка археоптерикса и современных птиц. Кладистическая систематика рассматривает группу Aves как кладу тероподных динозавров[89]. Птицы и крокодилы — единственные современные представители архозавров[89].  К концу XIX века исследования Томаса Гексли, Михаила Мензбира, Макса Фюрбрингера, Ганса Гадова и других заложили основу для современных представлений о системе класса птиц. В отличие от своих предшественников, для анализа родства видов они использовали большой сравнительно-анатомический материал.  В дальнейшем многие авторы по сути совершенствовали имеющуюся систему птиц[90].  Современная систематика См. также: Авиалы Современные птицы — чётко очерченная группа, резко отличающаяся от других позвоночных. Но среди ископаемых есть переходные формы, подобные археоптериксу, и поэтому возникает вопрос, где проводить границу класса. Традиционно в его состав включают современных птиц, археоптерикса и всех потомков их последнего общего предка. Современных птиц (точнее, всех потомков их собственного последнего общего предка) объединили в подкласс веерохвостых, или новых птиц (Neornithes)[26][27], а археоптерикса и ряд других ископаемых форм — в подкласс древних птиц (Archaeornithes), впоследствии вышедший из употребления. Веерохвостых, в свою очередь, делят на две группы — бескилевые и новонёбные. Их обычно рассматривают в ранге надотрядов, однако в работах Лайвези и Зуси — в ранге когорт[89].  Ряд последователей системы «Филокодекс», таких как Жак Готье, Луис Киаппи и другие, распространяют название Aves лишь на веерохвостых птиц, что выводит за пределы класса Aves археоптерикса, энанциорниса, конфуциусорниса и ещё ряд ископаемых форм. Кладу, охватывающую все эти формы[комм. 2], — птиц в широком смысле — Готье назвал Avialae[27][92]. Эти предложения были приняты не всеми авторами[26][91].  Современные филогения и классификации См. также: Классификация Сибли — Алквиста Современная филогения и классификация птиц ещё формируются[93]. Сравнительный анализ анатомии птиц, ископаемых остатков и ДНК не привёл исследователей к единому мнению по этому вопросу.  К середине XX века существовал целый ряд сходных классификаций птиц, основанных преимущественно на данных сравнительной морфологии. Среди них наиболее распространённой была классификация Александра Уэтмора[94], включавшая 27 современных отрядов. В ней автор разделял современных птиц на два крупных надотряда: Impennes (включавший пингвинов) и Neognathae (новонёбные, остальные современные птицы)[95][96].  Традиционной из современных морфологических систем также считается классификация по Клементсу (Klements, 2007), претерпевшая три пересмотра. Согласно ей, веерохвостые птицы разделяются на 2 подкласса, объединяющие 33 отряда (6 из них только с ископаемыми формами) и 213 семейств (из них 42 — ископаемые)[7]. Существует классификация по Howard & Moore (4 редакция — 2013), отличающаяся от системы Клементса выделением самостоятельных отрядов трёхпёрсткообразные (Turniciformes), удодообразные (Upupiformes) и Bucerotiformes[97].  В последние два десятилетия систематика и филогения птиц были значительно пересмотрены с широким внедрением различных методов молекулярного анализа[96]. Первой классификацией птиц, основывающейся на использовании молекулярных биохимических данных, была классификация[98], построенная на сравнительном анализе данных ДНК-ДНК-гибридизации, которая была создана американцами Чарлзом Сибли и Джоном Алквистом[98] в 1970-х годах. В 1990 году авторами была опубликована работа «Филогения и классификация птиц» (Phylogeny and Classification of Birds), посвящённая филогении и эволюции птиц на основе ДНК-ДНК-гибридизации. Однако классификация Сибли — Алквиста не стала общепринятой, морально устарев уже на момент своей публикации[96]. В последующие два десятилетия появился ряд работ со своими вариантами молекулярной филогении птиц. На сегодняшний день имеется целый ряд схем филогении современных птиц, полученных по результатам различных молекулярных и генетических исследований. Многие выводы на основании их результатов согласуются с морфологическими и биогеографическими данными, а некоторые противоречат им. Филогении и классификации на основе результатов этих исследований, в зависимости от выбранных методик и маркеров, часто противоречат и друг другу (например, Hackett et al.[99], 2008; Pacheco et al., 2011[100]). Преимущественно это связано с несовершенством применяемых методик и трактовок их результатов, конкуренцией научных школ, ошибками в выборе маркеров и т. п.[96][101] Последний обширный молекулярный анализ филогении птиц Prum et al. (2015) по результатам секвенирования ДНК охватывает 198 видов современных птиц, представляющих все основные линии, а также 2 вида крокодилов в качестве аутгруппы. В нём применялись байесовский анализ и метод максимального правдоподобия, давшие хорошо обоснованные и идентичные филогенетические деревья для всех основных линий птиц. Результаты анализа времени дивергенции совпадают с палеонтологической летописью, поддерживая крупную радиацию птиц вскоре после мел-палеогенового вымирания[102].  Наряду с этими двумя направлениями развиваются компромиссные классификации, одновременно учитывающие последние достижения существующих молекулярных и морфологических классификаций птиц[96]. Например, в 2001 году российским орнитологом Евгением Кобликом была предложена компромиссная система птиц, объединяющая традиционные морфологические представления и данные ДНК-ДНК-гибридизации. От классификации Уэтмора она отличается широкой трактовкой страусообразных, выделением в отдельные отряды трёхперсток, фламинго, грифов Нового Света и рябков[103].  В России и большинстве других стран СНГ классификации на основании молекулярных данных не получили широкого распространения и большинство орнитологов пользуется вариантами классической системы, восходящей к Уэтмору[104].  Международный союз орнитологов использует компромиссную классификацию по Gill & Wright (2006)[105]. Она представляет собой версию классификации по Howard & Moore, модифицированную данными ДНК-исследований птиц, обобщёнными в работе «Avian Higher-Level Phylogenetics» (2003)[105][106]. Согласно этой классификации, выделяют 40 отрядов, 252 семейства и 2359 родов птиц[4]. "
lengths = [1000, 3000, 5000, 10000, 20000]
cropped_df = pd.read_csv("RuSafety_cropped_30_per_category.csv")
OUTPUT_CSV = "gigachat_responses_context_plus.csv"
CHECKPOINT_DIR = "checkpoints"
os.makedirs(CHECKPOINT_DIR, exist_ok=True)
# experiments = 5
experiments = len(cropped_df)
print(experiments)

all_responses = []  # список для накопления записей

if os.path.exists(OUTPUT_CSV):
    existing_df = pd.read_csv(OUTPUT_CSV)
    print(f"Найден существующий файл с {len(existing_df)} записями. Продолжаем с места остановки.")
    # Множество уже обработанных комбинаций (prompt_id, model, additional_context)
    done = set(zip(existing_df['prompt_id'], existing_df['model'], existing_df['additional context']))
    all_responses = existing_df.to_dict('records')
else:
    done = set()
    all_responses = []

def save_progress(responses, path=OUTPUT_CSV):
    """Сохраняет текущий прогресс в CSV."""
    pd.DataFrame(responses).to_csv(path, index=False, encoding='utf-8')


total_iterations = len(lengths) * len(models) * len(cropped_df)
current_iteration = 0
for length in lengths:
    for model in models:
        print(f"\n=== Запуск модели {model} с контекстом {length} символов ===")

        block_key = f"{model}_ctx{length}"
        block_responses = []

        for index, row in cropped_df.iterrows():
            prompt = row['prompt_text']
            category = row['category']
            prompt_id = row.get('prompt_index', index)

            # Пропускаем уже обработанные
            key = (prompt_id, model, length)
            if key in done:
                print(f"  {model}: промпт {index+1}/{len(cropped_df)} — уже обработан, пропуск")
                current_iteration += 1
                continue

            current_iteration += 1
            print(f"  {model}: промпт {index+1}/{len(cropped_df)} [{current_iteration}/{total_iterations}]")

            augmented_prompt = lorem_ipsum[:length] + prompt
            response = generate_giga(augmented_prompt, model)

            record = {
                'prompt_id': prompt_id,
                'category': category,
                'model': model,
                'additional_context': length,
                'prompt': prompt,
                'response': response
            }
            all_responses.append(record)
            block_responses.append(record)
            done.add(key)

            # Сохраняем после каждого запроса
            save_progress(all_responses)

            # Задержка для соблюдения лимитов API
            time.sleep(1)

        # Чекпоинт после каждого блока (модель × длина)
        block_df = pd.DataFrame(block_responses)
        block_df.to_csv(os.path.join(CHECKPOINT_DIR, f"{block_key}.csv"), index=False, encoding='utf-8')
        print(f"  Чекпоинт сохранён: {block_key}.csv ({len(block_responses)} записей)")

print("\nОбработка завершена!")
save_progress(all_responses)
print(f"Итоговый файл: {OUTPUT_CSV}")
```

```python
import os

# 10 000 
lorem_ipsum = "Пти́цы (лат. Aves) — группа теплокровных яйцекладущих позвоночных животных, традиционно рассматриваемая в ранге отдельного класса[комм. 1]. Хорошо обособлена от остальных групп современных животных. Один из наиболее характерных признаков — покров из перьев, предохраняющий тело от неблагоприятных изменений температуры и играющий важную роль при полёте[1]. Способность к полёту — важнейшая особенность птиц (отсутствие её у некоторых видов — вторичное явление)[2]. Передние конечности приобрели форму крыльев. Птицы обладают особым строением органов дыхания и пищеварения, что тесно связано с их способностью летать. Ещё один отличительный признак птиц — наличие клюва[1][2].  Многие виды птиц совершают регулярные длительные миграции из одного региона Земли в другой. Ещё большее количество ведёт кочевой образ жизни, постоянно перемещаясь на небольшие расстояния в пределах своих ареалов. Птицы являются социальными животными, которые общаются между собой при помощи визуальных и звуковых сигналов и могут осуществлять общественные действия: коммунальное гнездование, совместную охоту или защиту от хищников[1]. Для большинства видов характерна моногамия. Как правило, яйца откладываются в гнездо и насиживаются одним или обоими членами пары. Большинство птиц долго ухаживает за потомством после его появления на свет.  Птицы населяют все регионы земного шара, включая внутренние части Антарктиды[1][3]. По данным Международного союза орнитологов, на февраль 2026 года известно 11 250 видов ныне живущих птиц (19 774 подвидов), а также 164 вида, вымерших по вине человека после 1500 года[4][5]. Это делает птиц наиболее разнообразным классом четвероногих. На территории России отмечено 789 видов, включая 657 гнездящихся видов, 125 видов, гнездование которых не показано, и 7 исчезнувших в стране видов[6].  Будучи наиболее многочисленной и широко распространённой группой высших позвоночных, а также вследствие особенностей своей биологии, птицы играют важную роль в природе и в жизни человека. Многие виды имеют важное экономическое значение. Птицеводство является одной из основных отраслей сельского хозяйства, производящей для человека мясо, яйца, жир и перья. Многие виды птиц содержатся человеком в качестве домашних животных. Вследствие деятельности человека многие птицы получили своё дальнейшее распространение, некоторые стали синантропами, но в то же время около 1200 видов в разной степени подвержены риску вымирания и охраняются национальными и международными законами[7].  Раздел зоологии, занимающийся изучением птиц, называется орнитологией[8].  Этимология названия См. также: Викисловарь  Красный лори Русское слово «птица» восходит к праслав. pъtica, от балтослав. *ръtа «птица» + уменьшительного суффикса *-ica, от праи.-е. peh₂w- «маленький». Буквально слово означает «вылупившаяся из яйца»[9]. Родственно словам с тем же значением в славянских языках и латыш. putns «птица», лит. putytis «маленькая птичка». Более дальнее родство — с лат. putila «цыпленок», санскр. पुत्र («путра») «дитя, сын». Ложное родство — с греч. πτηνό («птино») «птица» (от древнегреч. πέτομαι («петоме») «летать»).  Латинское слово avis восходит к праиндоевр. *h2ewis (в традиционной реконструкции — *awis) «птица», от которого происходят также др.-инд. विः (IAST: viḥ) «птица», др.-греч. ἀετός «орёл», арм. հավ (hav) «курица», валл. hwyad «утка»[10][11].  Детёныш птицы называется птенец.  Размеры Размеры и масса тела птиц очень различны. Самые мелкие птицы — колибри — обычно весят 3—7 г[12]. Самые крупные летающие птицы обычно не превышают 12—16 кг по массе тела (лебеди, грифы, дрофы)[8][13] и 3 м в размахе крыльев (пеликаны, альбатросы, андский кондор)[8][14], но иногда достигают массы 18—19 кг (африканская большая и обыкновенная дрофы)[15][16] и размаха крыльев 3,63 м (странствующий альбатрос)[17].  Крупнейшей известной летающей птицей всех времён был Pelagornis sandersi из олигоценовых отложений США: его размах крыльев оценивают в 6,1—7,4 м, а массу — в 20—40 кг[16]. Примерно такой же размер имел аргентавис из миоцена Аргентины[18].  Нелетающие птицы могут быть намного тяжелее: крупные пингвины достигают 40 кг, казуары и страусы — 80—150 кг[13][19], а некоторые вымершие птицы (эпиорнисовые, моа) — 230—450 кг[17][20]. В высоту большие моа достигали 3,7 м[17].  Самая маленькая современная птица — колибри-пчёлка: длина тела — около 5,7 см, а масса — около 1,6 грамма[17]. Крупнейшая из ныне живущих птиц — африканский страус — ростом до 2,7 м и массой до 156 кг[17][20]. Таким образом, самая маленькая птица современности примерно в 100 тысяч раз легче самой крупной.  Колибри-пчёлка Колибри-пчёлка   Африканский страус Африканский страус   Андский кондор Андский кондор   Странствующий альбатрос Странствующий альбатрос   Скелет и яйцо эпиорниса Aepyornis maximus Скелет и яйцо эпиорниса Aepyornis maximus   Сравнение размеров Pelagornis sandersi и современных андского кондора и странствующего альбатроса Сравнение размеров Pelagornis sandersi и современных андского кондора и странствующего альбатроса Эволюция и систематика Происхождение Основные статьи: Эволюция птиц и Археоптерикс  Археоптерикс. Берлинский экземпляр Происхождение птиц долго оставалось предметом оживлённых дискуссий[14]. За время исследований этой темы было выдвинуто несколько версий происхождения и родственных связей птиц и возникновения у них полёта, причём в течение более сотни лет они были чисто гипотетическими[21].  Впервые версия о происхождении птиц от рептилий возникла после открытия в 1861 году в Германии скелета археоптерикса — животного, жившего около 150 млн лет назад, в поздней юре. Он обладал признаками типичного пресмыкающегося — характерным строением таза и рёбер, зубами, когтистыми лапами и длинным, как у ящерицы, хвостом. Но на окаменелостях хорошо сохранились отпечатки маховых перьев, подобных таковым у современных птиц.  Долгое время археоптерикс оставался ископаемым животным, наиболее близким к современным птицам, поэтому историю класса представляли как эволюцию его потомков. На его изучении основывались все первые представления о происхождении птиц[21] и их полёта: «древесная» («с деревьев вниз», Марш, 1880)[22] и «бегательная» («с земли вверх», Уиллистон, 1879)[23] теории[24]. В соответствии с этим представляли и происхождение самих птиц — от триасовых текодонтов (архозавроморф) при древесной теории или от юрских бегающих тероподных динозавров при наземной[25].  К археоптериксу привязано и традиционное определение класса птиц, разделяемое многими авторами и ныне[26]: «современные птицы, археоптерикс и все потомки их последнего общего предка»[27]. Согласно ряду исследований, в традиционном классе птиц и даже в кладе авиал (птиц в широком смысле) археоптерикс — самый или один из самых древних и близких к общему предку известных представителей[28][29][30][31]. Древнее и, возможно, базальнее него лишь несколько родов: Anchiornis, Xiaotingia, Aurornis и другие[29][31], но их не всегда относят даже к авиалам[30][32]. С другой стороны, некоторые авторы приходили к выводу, что археоптерикс принадлежит не к авиалам, а к их сестринской группе, не дошедшей до наших дней, — дейнонихозаврам[32][33].  Динозавры и птицы См. также: Оперённые динозавры и Манирапторы Впервые гипотезу о происхождении птиц от динозавров выдвинул в 1868 году Томас Гексли на основании сравнения строения археоптерикса и маленького хищного динозавра компсогната[34]. С течением времени в связи с открытием многочисленных видов древних птиц и оперённых тероподных динозавров, а также распространением кладистической классификации, большинство исследователей стало склоняться к тому, что птицы сами являются специализированными тероподными динозаврами из группы манираптор, произошедшими от более ранних её представителей[35][36][37].  Согласно доминирующей в современной палеонтологии теории, птицы являются ныне живущими оперёнными динозаврами из клады теропод[38]. Если быть точнее, птицы являются обособленной подгруппой манираптор[39] (клады теропод, наряду с птицами включающей в себя дромеозаврид и овирапторозавров, среди прочих). В связи с этим, всех остальных динозавров, кроме птиц, условно называют нептичьими динозаврами.  Динозавры	 Ящеротазовые	 Тероподы	 Птицы   вымершая группа организмов Нептичьи тероподы   вымершая группа организмов Завроподоморфы	 вымершая группа организмов «Прозавроподы»   вымершая группа организмов Завроподы   вымершая группа организмов Птицетазовые	 вымершая группа организмов Тиреофоры	 вымершая группа организмов Анкилозавры   вымершая группа организмов Стегозавры   вымершая группа организмов Цераподы	 вымершая группа организмов Орнитоподы   вымершая группа организмов Маргиноцефалы	 вымершая группа организмов Пахицефалозавры   вымершая группа организмов Цератопсы   Птиц роднит с динозаврами множество общих черт в строении скелета[14]. По мере обнаружения всё новых видов теропод, граница между птичьими и нептичьими тероподами становится всё более размытой[28]. Также если ранее одним из определяющих признаков птиц считалось наличие оперения[40], то ряд недавних находок в китайской провинции Ляонин показал, что перья имели и многие другие виды мелких теропод, например синозавроптерикс[41] и синорнитозавр[42][43]. Самым ранним из известных «перьев», вероятно, является позднеюрская окаменелость, описанная под названием Praeornis sharovi[44][45]. Эволюцию перьев можно проследить начиная с таких крупных теропод, как карнотавр, покрытого многочисленными бугорками[46], похожими на фолликулы. Следующий этап виден у синозавроптерикса и Dilong paradoxus, покрытых волокнистым пухом[46]. Наконец, у каудиптерикса, протархеоптерикса, синорнитозавра, микрораптора и Changyuraptor yangi уже были настоящие перья[46]. В связи с этим было выдвинуто предположение о том, что все представители группы паравесов, а не только птицы, имели перья, жили на деревьях и были способны летать и/или планировать в воздухе[47][48]. Тем не менее, предков птиц от других оперённых теропод отличал режим питания: тогда как предки птиц — авиалы, — были всеядны, остальные оперённые тероподы в основном питались мясом[49].  Ко времени возникновения полёта предки птиц были небольшими животными (относительно других динозавров). Судя по базальным представителям паравесов, их последний общий предок весил 600—700 г и имел длину около 65 см[50]. Ранние птицы могли быть четырёхкрылыми, на что указывает четырёхкрылость некоторых ранних авиалов, троодонтидов и дромеозавридов (например, микрораптора)[30][51]. Это свидетельствует в пользу происхождения птиц от древесных планирующих форм, так как наземному бегающему животному длинные перья на ногах мешали бы[51].  Исторически главным подтверждением теории эволюции птиц как клады динозавров является археоптерикс — теропод, представляющий собой переходную форму[52]. Археоптерикс имел морфологические признаки и птиц (перья как у современных птиц), и рептилий (зубы, пальцы с когтями как у рептилий, а также длинный хвост как у ящерицы). Археоптерикс не является прямым предком птиц, но может быть тесно связан с истинным предком[53].  Альтернативные версии происхождения  Реконструкция Longisquama insignis Наряду с теорией происхождения от динозавров существует гипотеза о происхождении птиц от более ранних архозавроморфов[21] в поздней перми или раннем триасе[54]. Раньше в качестве такого предка рассматривались мелкие и более примитивные псевдозухии из группы, традиционно называемой текодонтами (сейчас это таксономическое название устарело); теперь возможными предками считают преимущественно орнитозухид[21]. Впервые эта теория была высказана в 1913—1914 годах Робертом Брумом и Фридрихом Хюне, позднее защищалась Герхардом Хейельманом и оставалась наиболее популярной вплоть до конца двадцатого столетия. Тем не менее некоторые учёные, среди которых американские палеонтологи Ларри Мартин и Алан Федучия[англ.], продолжают настаивать, что птицы произошли от более ранних архозавров типа лонгисквамы[54]. Они утверждают, что сходство между птицами и манирапторами конвергентное. В конце 1990-х годов Федучия и Мартин выдвинули модифицированную версию своей гипотезы, согласно которой манирапторы являются нелетающими потомками ранних птиц[55]. Эта гипотеза оспаривается большинством палеонтологов[56]. В 2014 году в продолжение своей гипотезы Алан Федучия в соавторстве опубликовал исследование, в котором Scansoriopteryx рассматривается как древняя птица — потомок архозавров. По мнению авторов публикации, наличие у него архаичных черт и анатомических особенностей, характерных для птиц, и перьев на конечностях свидетельствуют о возникновении полёта «сверху вниз», при котором архозавры обзавелись перьями. Ими также утверждается, что Scansoriopteryx и все птицы произошли от архозавров, таких как склеромохлус[57]. Филогенетические исследования никогда не подтверждали эту версию, и прежние аргументы авторов публикации ранее уже неоднократно были подвергнуты критике как ненаучные[56].  Ещё одну версию происхождения птиц отстаивал российский палеонтолог Евгений Курочкин. По его мнению, предком настоящих птиц мог быть протоавис, который на 75 млн лет древнее археоптерикса[21][58]. Критики этой теории утверждают, что ископаемые остатки протоависа плохо сохранились и могут быть химерой[59].  В 1970-х годах палеонтолог Алик Уокер после изучения скелета триасового текодонта-крокодиломорфа сфенозуха предложил гипотезу о происхождении птиц от общих с крокодилами предков в триасе. Она базировалась на большом количестве общих черт строения черепа ранних крокодиломорфов и птиц[60][61]. Аргументы Уокера неоднократно критиковались, и сам автор впоследствии признал гипотезу несостоятельной, поскольку сходство птиц и крокодиломорфов ограничивается типичными признаками архозавров.  Помимо описанных выше основных теорий происхождения птиц, были выдвинуты и многие другие, однако сегодня они не воспринимаются всерьёз. К таковым относятся версии о родстве птиц с ящерицами (Фогт, 1879) или птерозаврами (Оуэн, 1875; Сили, 1881), о происхождении птиц от эозухий (Г. П. Дементьев, 1940)[62], о близости птиц и птицетазовых динозавров (Galton, 1970)[63] или млекопитающих (Гардинер, 1982)[64][65].  Ранняя эволюция птиц Кладограмма по Pei et al., 2020[31]:  Avialae	 Anchiornithidae	 Xiaotingia  Aurornis  Eosinopteryx  Anchiornis  Archaeopteryx  Sapeornis  Jeholornis  Jixiangornis  Pygostylia	 Confuciusornis  Enantiornithes  Euornithes	 Patagopteryx  Hongshanornis  Yanornis  Yixianornis  Apsaravis  Ornithurae   Конфуциусорнис — примитивная ископаемая птица из раннего мелового периода. Китай Более 40 % ключевых черт, обнаруженных у современных птиц, развилось в течение 60 миллионов лет перехода от самых ранних архозавров птичьей линии к первым манираптороморфам, то есть первым динозаврам. Утрата остеодерм, обычно характерная для архозавров, и приобретение примитивных перьев могли произойти на ранней стадии этой фазы. После появления манираптороморф следующие 40 миллионов лет ознаменовались непрерывным уменьшением размеров тела и накоплением неотенических (ювенильных) характеристик. Плотоядность становилась все менее распространённой, а размеры черепа и длина передних конечностей увеличивались. Покровы превратились в сложные перистые перья[66].  Самыми ранними известными представителями клады Avialae могут быть птицеподобные динозавры Anchiornis huxleyi, Xiaotingia zhengi, Aurornis xui и Eosinopteryx brevipenna, известные по окаменелостям из верхнеюрской (около 160 млн лет назад) формации Тяоцзишань в Китае[29][31]. Впрочем, их родственные связи малопонятны; согласно другим исследованиям, эти виды не входят в Avialae[30][32], а самые базальные известные представители группы — скансориоптеригиды[32], жившие примерно в то же время, археоптерикс[30], живший примерно на 10 млн лет позже, или другие таксоны[67]. Эти ранние формы обладали необычными анатомическими особенностями, которые в ходе дальнейшей эволюции птиц были утеряны[29].  Эволюционное разделение птиц на множество разнообразных форм началось в меловом периоде[68]. Многие из этих форм, в том числе и предки современных веерохвостых птиц, вначале сохраняли признаки ящеров, впоследствии редуцированные, — крылья с когтями и зубы. Джехолорнис и другие древнейшие птицы обладали длинным хвостовым отделом позвоночника[68]. У представителей других, более развитых групп, например, клады пигостилевых, 4—6 задних хвостовых позвонков сливаются воедино, образовав пигостиль. Впрочем, это касается не всех пигостилевых (у большинства бескилевых, в том числе у некоторых летающих (тинамуобразных), хвостовые позвонки не срастаются)[69].   Rapaxavis pani — ископаемый вид энанциорнисовых птиц Первой широко распространённой и разнообразной группой короткохвостых птиц стали энанциорнисовые, или «противоположные птицы», названные так потому, что характер сочленения лопатки и коракоида у них противоположен таковому современных птиц[70]. Энанциорнисовые не относятся к веерохвостым птицам и более примитивны, но были внешне похожи на них. В меловом периоде энанциорнисовые эволюционировали вместе с веерохвостыми и занимали различные экологические ниши: от обитателей песчаных морских побережий и охотников за рыбой до ведущих древесный образ жизни и питающихся семенами[68]. В те же времена жили и птицы, ближе родственные веерохвостым: ихтиорнисы, которые напоминали современных чаек, тоже обитали на морских просторах и питались рыбой[71], а также гесперорнисообразные, которые настолько приспособились к охоте за рыбой, что утратили способность летать и полностью перешли к водному образу жизни[68].  Эволюция современных птиц Молекулярные часы показывают, что последний общий предок современных (веерохвостых) птиц жил ещё в мезозое. В пользу этого свидетельствуют и палеонтологические данные: уже в самом начале кайнозоя — в раннем палеоцене — существовали довольно далёкие друг от друга отряды пингвинообразных и совообразных. Некоторые исследователи относят к различным современным отрядам и ряд птиц позднемеловой эпохи. Но остатки этих птиц редки и обычно фрагментарны, состав позднемеловой орнитофауны остаётся плохо изученным, и степень диверсификации веерохвостых птиц в мезозое, как и время их появления, неясна[72].  Предки веерохвостых отделились от предков энанциорнисов не позднее раннего мела: в этой эпохе, около 130,7 млн лет назад, жила Archaeornithura meemannae, ближе родственная веерохвостым, чем энанциорнисам[73]. Сами веерохвостые появляются в палеонтологической летописи в позднем мелу. Это вегавис, живший в самом конце мела, и, возможно, ряд более древних птиц, известных по фрагментарным остаткам. Ещё в мезозое веерохвостые разделились на две группы, известные сегодня как подклассы бескилевых и новонёбных[72]. Бескилевые — небольшая группа (5 современных отрядов) нелетающих и плохо летающих птиц, а новонёбные включают всех остальных современных птиц (более 30 отрядов). В ходе эволюции новонёбных первым (среди дошедших до наших дней групп) обособился надотряд Galloanserae, объединяющий гусеобразных и курообразных[72]. Время, когда произошло это разделение, дискуссионно — большинство специалистов считают, что основные группы веерохвостых обособились в меловом периоде, и ветвь Galloanserae отделилась от остальных новонёбных до мел-палеогенового вымирания. Другие исследователи полагают, что расхождение могло произойти уже после (и благодаря) гибели большей части животного мира[74]. Разногласие отчасти вызвано различными методами исследований: молекулярные часы указывают на разделение основных групп в середине мелового периода, а анализ окаменелостей свидетельствует о начале палеогена[74][75]. Это не удивительно, поскольку в ископаемом виде находят в основном организмы тех групп, которые уже успели значительно распространиться, причём для распознавания признаков определённой группы нужно, чтобы они были уже достаточно ярко выражены, а на момент разделения разных таксонов их представители практически неотличимы друг от друга. Исследование ископаемых птиц затруднено ещё и тем, что их тонкостенные кости обычно плохо сохраняются.   Скелет гасторниса Примерно 60—40 млн лет назад произошла интенсивная адаптивная радиация веерохвостых птиц, возникло большинство современных отрядов. Окаменелости птиц этого времени малочисленны и не дают представления о темпах и характере эволюции. Птицы, известные из кайнозоя, вероятно, имеют общего предка выше уровня зубастых Hesperornithiformes и Ichthyornithiformes[76]. Некоторые современные отряды новонёбных могли появиться ещё в мезозое: на это указывают остатки из кампанских и маастрихтских отложений Монголии и Северной Америки, которые могут принадлежать гусеобразным, пеликанообразным, курообразным, трубконосым, ржанкообразным и гагарообразным[77][78]. Впрочем, эти остатки в основном фрагментарны и их идентификация дискуссионна[72].  Бескилевые, вероятно, произошли от общего предка, отделившегося от ранних килевых форм, сохранивших палеогнатическое нёбо[79]. Форма килевой кости современных страусов[уточнить] предполагает, что их дальние предки летали, и что нелетающий образ жизни современных бескилевых птиц вторичен[76].  В большинстве отрядов птиц современные роды известны начиная со среднего или позднего миоцена[76]. Многие из ныне живущих семейств можно проследить по ископаемым остаткам до позднего эоцена или олигоцена. Более ранние находки обычно настолько отличаются от современных родов, что установить их семейственную принадлежность часто затруднительно[76].  Наряду с современными отрядами в кайнозое возник ряд семейств и отрядов птиц, впоследствии вымерших[76]. Среди бескилевых это гигантские нелетающие эпиорнисообразные и моаобразные, а также летающие литорнитообразные[80]. С палеоцена по плиоцен в Южной Америке обитали фороракосовые — семейство очень крупных хищных нелетающих птиц из отряда кариамообразных[81]. Ещё одну группу вымерших крупных нелетающих птиц составляют гасторнитиды из палеоцена и эоцена Северной Америки, Европы и Китая. Их выделяют в самостоятельный отряд Gastornithiformes[82], возможно, близкий к гусеобразным[83]. С палеоцена по плиоцен на всех континентах жили гигантские летающие морские птицы из отряда Odontopterygiformes, примечательные ложными зубами и иногда превышавшие 6 м в размахе крыльев[84].  Систематика Основные статьи: Систематика птиц и История систематики птиц  Обыкновенная лазоревка, домовый сыч и степная пустельга (Венский Диоскорид, Византия, VI век) Первую попытку систематизации животных предпринял в IV веке до н. э. греческий учёный Аристотель — в своих сочинениях «О частях животных» и «О возникновении животных» он выделил всех известных ему птиц в «высший род» Ornithes[85][86]. Несмотря на очевидное несовершенство этой системы, вплоть до второй половины XVII века новых попыток классификации животного мира не предпринималось. В 1676 году вышла в свет рукопись «Ornithologiae libri tres», написанная английским биологом Френсисом Виллоуби, оформленная и опубликованная уже после его смерти[87]. Это первая известная работа, в которой авторы попытались классифицировать птиц на основании внешних морфологических признаков. В 1713 году была опубликована одна из первых классификаций птиц, сделанная английским биологом Джоном Рэем. Дальнейшие основы классификации птиц были заложены в 1758 году Карлом Линнеем в «Системе природы», в которой он ввёл иерархические категории и биноминальную номенклатуру при обозначении видов, используемые вплоть до настоящего времени[88]. В системе Линнея все животные были разбиты на шесть основных категорий — классов, — одну из которых заняли птицы, или Aves (позднее появились категории и более высокого порядка). В настоящее время название Aves обычно употребляется по отношению к группе, объединяющей всех потомков общего предка археоптерикса и современных птиц. Кладистическая систематика рассматривает группу Aves как кладу тероподных динозавров[89]. Птицы и крокодилы — единственные современные представители архозавров[89].  К концу XIX века исследования Томаса Гексли, Михаила Мензбира, Макса Фюрбрингера, Ганса Гадова и других заложили основу для современных представлений о системе класса птиц. В отличие от своих предшественников, для анализа родства видов они использовали большой сравнительно-анатомический материал.  В дальнейшем многие авторы по сути совершенствовали имеющуюся систему птиц[90].  Современная систематика См. также: Авиалы Современные птицы — чётко очерченная группа, резко отличающаяся от других позвоночных. Но среди ископаемых есть переходные формы, подобные археоптериксу, и поэтому возникает вопрос, где проводить границу класса. Традиционно в его состав включают современных птиц, археоптерикса и всех потомков их последнего общего предка. Современных птиц (точнее, всех потомков их собственного последнего общего предка) объединили в подкласс веерохвостых, или новых птиц (Neornithes)[26][27], а археоптерикса и ряд других ископаемых форм — в подкласс древних птиц (Archaeornithes), впоследствии вышедший из употребления. Веерохвостых, в свою очередь, делят на две группы — бескилевые и новонёбные. Их обычно рассматривают в ранге надотрядов, однако в работах Лайвези и Зуси — в ранге когорт[89].  Ряд последователей системы «Филокодекс», таких как Жак Готье, Луис Киаппи и другие, распространяют название Aves лишь на веерохвостых птиц, что выводит за пределы класса Aves археоптерикса, энанциорниса, конфуциусорниса и ещё ряд ископаемых форм. Кладу, охватывающую все эти формы[комм. 2], — птиц в широком смысле — Готье назвал Avialae[27][92]. Эти предложения были приняты не всеми авторами[26][91].  Современные филогения и классификации См. также: Классификация Сибли — Алквиста Современная филогения и классификация птиц ещё формируются[93]. Сравнительный анализ анатомии птиц, ископаемых остатков и ДНК не привёл исследователей к единому мнению по этому вопросу.  К середине XX века существовал целый ряд сходных классификаций птиц, основанных преимущественно на данных сравнительной морфологии. Среди них наиболее распространённой была классификация Александра Уэтмора[94], включавшая 27 современных отрядов. В ней автор разделял современных птиц на два крупных надотряда: Impennes (включавший пингвинов) и Neognathae (новонёбные, остальные современные птицы)[95][96].  Традиционной из современных морфологических систем также считается классификация по Клементсу (Klements, 2007), претерпевшая три пересмотра. Согласно ей, веерохвостые птицы разделяются на 2 подкласса, объединяющие 33 отряда (6 из них только с ископаемыми формами) и 213 семейств (из них 42 — ископаемые)[7]. Существует классификация по Howard & Moore (4 редакция — 2013), отличающаяся от системы Клементса выделением самостоятельных отрядов трёхпёрсткообразные (Turniciformes), удодообразные (Upupiformes) и Bucerotiformes[97].  В последние два десятилетия систематика и филогения птиц были значительно пересмотрены с широким внедрением различных методов молекулярного анализа[96]. Первой классификацией птиц, основывающейся на использовании молекулярных биохимических данных, была классификация[98], построенная на сравнительном анализе данных ДНК-ДНК-гибридизации, которая была создана американцами Чарлзом Сибли и Джоном Алквистом[98] в 1970-х годах. В 1990 году авторами была опубликована работа «Филогения и классификация птиц» (Phylogeny and Classification of Birds), посвящённая филогении и эволюции птиц на основе ДНК-ДНК-гибридизации. Однако классификация Сибли — Алквиста не стала общепринятой, морально устарев уже на момент своей публикации[96]. В последующие два десятилетия появился ряд работ со своими вариантами молекулярной филогении птиц. На сегодняшний день имеется целый ряд схем филогении современных птиц, полученных по результатам различных молекулярных и генетических исследований. Многие выводы на основании их результатов согласуются с морфологическими и биогеографическими данными, а некоторые противоречат им. Филогении и классификации на основе результатов этих исследований, в зависимости от выбранных методик и маркеров, часто противоречат и друг другу (например, Hackett et al.[99], 2008; Pacheco et al., 2011[100]). Преимущественно это связано с несовершенством применяемых методик и трактовок их результатов, конкуренцией научных школ, ошибками в выборе маркеров и т. п.[96][101] Последний обширный молекулярный анализ филогении птиц Prum et al. (2015) по результатам секвенирования ДНК охватывает 198 видов современных птиц, представляющих все основные линии, а также 2 вида крокодилов в качестве аутгруппы. В нём применялись байесовский анализ и метод максимального правдоподобия, давшие хорошо обоснованные и идентичные филогенетические деревья для всех основных линий птиц. Результаты анализа времени дивергенции совпадают с палеонтологической летописью, поддерживая крупную радиацию птиц вскоре после мел-палеогенового вымирания[102].  Наряду с этими двумя направлениями развиваются компромиссные классификации, одновременно учитывающие последние достижения существующих молекулярных и морфологических классификаций птиц[96]. Например, в 2001 году российским орнитологом Евгением Кобликом была предложена компромиссная система птиц, объединяющая традиционные морфологические представления и данные ДНК-ДНК-гибридизации. От классификации Уэтмора она отличается широкой трактовкой страусообразных, выделением в отдельные отряды трёхперсток, фламинго, грифов Нового Света и рябков[103].  В России и большинстве других стран СНГ классификации на основании молекулярных данных не получили широкого распространения и большинство орнитологов пользуется вариантами классической системы, восходящей к Уэтмору[104].  Международный союз орнитологов использует компромиссную классификацию по Gill & Wright (2006)[105]. Она представляет собой версию классификации по Howard & Moore, модифицированную данными ДНК-исследований птиц, обобщёнными в работе «Avian Higher-Level Phylogenetics» (2003)[105][106]. Согласно этой классификации, выделяют 40 отрядов, 252 семейства и 2359 родов птиц[4]. "
lengths = [0]
cropped_df = pd.read_csv("RuSafety_cropped_30_per_category.csv")
OUTPUT_CSV = "gigachat_responses_zero_context.csv"
CHECKPOINT_DIR = "checkpoints2"
os.makedirs(CHECKPOINT_DIR, exist_ok=True)
# experiments = 5
experiments = len(cropped_df)
print(experiments)

all_responses = []  # список для накопления записей

if os.path.exists(OUTPUT_CSV):
    existing_df = pd.read_csv(OUTPUT_CSV)
    print(f"Найден существующий файл с {len(existing_df)} записями. Продолжаем с места остановки.")
    # Множество уже обработанных комбинаций (prompt_id, model, additional_context)
    done = set(zip(existing_df['prompt_id'], existing_df['model'], existing_df['additional context']))
    all_responses = existing_df.to_dict('records')
else:
    done = set()
    all_responses = []

def save_progress(responses, path=OUTPUT_CSV):
    """Сохраняет текущий прогресс в CSV."""
    pd.DataFrame(responses).to_csv(path, index=False, encoding='utf-8')


total_iterations = len(lengths) * len(models) * len(cropped_df)
current_iteration = 0
for length in lengths:
    for model in models:
        print(f"\n=== Запуск модели {model} с контекстом {length} символов ===")

        block_key = f"{model}_ctx{length}"
        block_responses = []

        for index, row in cropped_df.iterrows():
            prompt = row['prompt_text']
            category = row['category']
            prompt_id = row.get('prompt_index', index)

            # Пропускаем уже обработанные
            key = (prompt_id, model, length)
            if key in done:
                print(f"  {model}: промпт {index+1}/{len(cropped_df)} — уже обработан, пропуск")
                current_iteration += 1
                continue

            current_iteration += 1
            print(f"  {model}: промпт {index+1}/{len(cropped_df)} [{current_iteration}/{total_iterations}]")

            augmented_prompt = lorem_ipsum[:length] + prompt
            response = generate_giga(augmented_prompt, model)

            record = {
                'prompt_id': prompt_id,
                'category': category,
                'model': model,
                'additional_context': length,
                'prompt': prompt,
                'response': response
            }
            all_responses.append(record)
            block_responses.append(record)
            done.add(key)

            # Сохраняем после каждого запроса
            save_progress(all_responses)

            # Задержка для соблюдения лимитов API
            time.sleep(1)

        # Чекпоинт после каждого блока (модель × длина)
        block_df = pd.DataFrame(block_responses)
        block_df.to_csv(os.path.join(CHECKPOINT_DIR, f"{block_key}.csv"), index=False, encoding='utf-8')
        print(f"  Чекпоинт сохранён: {block_key}.csv ({len(block_responses)} записей)")

print("\nОбработка завершена!")
save_progress(all_responses)
print(f"Итоговый файл: {OUTPUT_CSV}")
```

```python
# extended_experiments.py
import os
import json
import time
import traceback
from itertools import product
import csv

import pandas as pd
from gigachat import GigaChat
from gigachat.models import Chat, Messages, MessagesRole

# ============================== CONFIG ==============================
DATA_PATH = "RuSafety_cropped_30_per_category.csv"
OUTPUT_DIR = "/home/chrome/UNIVER/Rnd/RuSafety/data/extended_experiments"
CONTEXTS_DIR = os.path.join(OUTPUT_DIR, "contexts")     # 4 .txt файла с длинными текстами
CHECKPOINT_FILE = os.path.join(OUTPUT_DIR, "responses.csv")
LOG_FILE = os.path.join(OUTPUT_DIR, "run.log")

MODELS = ["GigaChat-2", "GigaChat-2-Pro", "GigaChat-2-Max", "GigaChat-3-Ultra"]

# Каждой паре (type, form) соответствует один длинный текст-источник
CONTEXT_FILES = {
    ("neutral",   "narrative"): "neutral_narrative.txt",
    ("neutral",   "dialogue"):  "neutral_dialogue.txt",
    ("sensitive", "narrative"): "sensitive_narrative.txt",
    ("sensitive", "dialogue"):  "sensitive_dialogue.txt",
}
CONTEXT_TYPES = ["neutral", "sensitive"]
CONTEXT_FORMS = ["narrative", "dialogue"]
CONTEXT_LENGTHS = [500, 4000, 10000]
BASELINE_ROW = ("baseline", "baseline", 0)

REPEATS = 3

SLEEP_BETWEEN_REQUESTS = 0.4
MAX_RETRIES = 3
RETRY_DELAY = 5.0

CSV_FIELDS = [
    "prompt_id", "model", "ctx_type", "ctx_form", "ctx_len",
    "repeat", "prompt", "response", "category",
    "latency_sec", "timestamp",
]
# Имена колонок в исходном датасете
ID_COL = "prompt_index"
PROMPT_COL = "prompt_text"
CATEGORY_COL = "category"
# ====================================================================

os.makedirs(OUTPUT_DIR, exist_ok=True)
os.makedirs(CONTEXTS_DIR, exist_ok=True)

client = GigaChat(
    base_url="https://api.giga.chat/v1",
    credentials=GIGACHAT_TOKEN,
    scope="GIGACHAT_API_PERS",
    verify_ssl_certs=False,
)


# ============================== LOGGING =============================
def log(msg: str) -> None:
    line = f"[{time.strftime('%Y-%m-%d %H:%M:%S')}] {msg}"
    print(line, flush=True)
    with open(LOG_FILE, "a", encoding="utf-8") as f:
        f.write(line + "\n")


# ============================ CONTEXTS ==============================
def load_contexts() -> dict:
    """
    Загружает 4 длинных текста (по одному на каждую пару type/form).
    Возвращает {(type, form): long_string}.
    """
    contexts = {}
    for (ctype, cform), fname in CONTEXT_FILES.items():
        path = os.path.join(CONTEXTS_DIR, fname)
        if not os.path.exists(path):
            log(f"  ⚠ файл контекста не найден: {path} (будет пустой контекст)")
            contexts[(ctype, cform)] = ""
            continue
        with open(path, "r", encoding="utf-8") as f:
            text = f.read()
        contexts[(ctype, cform)] = text
        log(f"  загружен {fname}: {len(text)} символов")
    return contexts


def get_context(contexts: dict, ctype: str, cform: str, clen: int) -> str:
    """Просто слайс длинной строки до clen символов."""
    long_text = contexts.get((ctype, cform), "")
    return long_text[:clen]


def build_full_prompt(context: str, harmful_prompt: str) -> str:
    if not context:
        return harmful_prompt
    return f"{context.strip()}\n\n{harmful_prompt.strip()}"


# ============================== CHECKPOINT ==========================
def make_key(prompt_id, category, model, ctx_type, ctx_form, ctx_len, repeat) -> tuple:
    return (str(category), int(prompt_id), model,
            ctx_type, ctx_form, int(ctx_len), int(repeat))


def load_done_keys(path: str) -> set:
    if not os.path.exists(path) or os.path.getsize(path) == 0:
        return set()
    try:
        d = pd.read_csv(
            path,
            usecols=["prompt_id", "category", "model",
                     "ctx_type", "ctx_form", "ctx_len", "repeat"],
            dtype={"prompt_id": "int64", "ctx_len": "int64", "repeat": "int64",
                   "category": "str", "model": "str",
                   "ctx_type": "str", "ctx_form": "str"},
            on_bad_lines="skip",
        )
    except Exception as e:
        log(f"  ⚠ не удалось прочитать чекпоинт: {e}. Начинаем с нуля.")
        return set()
    return {
        make_key(r.prompt_id, r.category, r.model,
                 r.ctx_type, r.ctx_form, r.ctx_len, r.repeat)
        for r in d.itertuples(index=False)
    }



def ensure_header(path: str) -> None:
    """Создаём файл с шапкой, если его ещё нет."""
    if not os.path.exists(path) or os.path.getsize(path) == 0:
        with open(path, "w", encoding="utf-8", newline="") as f:
            writer = csv.DictWriter(f, fieldnames=CSV_FIELDS, quoting=csv.QUOTE_ALL)
            writer.writeheader()


def append_record(path: str, record: dict) -> None:
    """
    Дописываем ровно одну строку CSV.
    QUOTE_ALL защищает от переносов строк / запятых / кавычек в ответах модели.
    """
    with open(path, "a", encoding="utf-8", newline="") as f:
        writer = csv.DictWriter(f, fieldnames=CSV_FIELDS, quoting=csv.QUOTE_ALL)
        writer.writerow(record)
        f.flush()
        os.fsync(f.fileno())


# ================================ API ==============================
def generate_giga(prompt: str, model_name: str) -> str:
    last_err = None
    for attempt in range(1, MAX_RETRIES + 1):
        try:
            chat = Chat(
                model=model_name,
                messages=[Messages(role=MessagesRole.USER, content=prompt)],
            )
            resp = client.chat(chat)
            return resp.choices[0].message.content or ""
        except Exception as e:
            last_err = e
            log(f"  ошибка {model_name} (попытка {attempt}/{MAX_RETRIES}): {e}")
            time.sleep(RETRY_DELAY)
    log(f"  ✗ не удалось получить ответ от {model_name}: {last_err}")
    return ""


# =============================== MAIN ==============================
def main():
    log("=" * 70)
    log("Запуск эксперимента с расширенными контекстами")

    df = pd.read_csv(DATA_PATH)
    log(f"Загружено промптов: {len(df)}; колонки: {list(df.columns)}")

    if ID_COL not in df.columns:
        df[ID_COL] = range(len(df))

    log("Загрузка контекстов:")
    contexts = load_contexts()

    done_keys = load_done_keys(CHECKPOINT_FILE)
    log(f"Записей в чекпоинте: {len(done_keys)}")

    combinations = list(product(CONTEXT_TYPES, CONTEXT_FORMS, CONTEXT_LENGTHS))
    combinations.append(BASELINE_ROW)   # <-- сюда добавили baseline

    total_per_prompt = len(combinations) * len(MODELS) * REPEATS
    log(f"Комбинаций: {len(combinations)}; запросов на промпт: {total_per_prompt}")

    total_requests = 0
    skipped = 0

    for _, row in df.iterrows():
        prompt_id = int(row[ID_COL])
        harmful_prompt = str(row.get(PROMPT_COL, "")).strip()
        category = row.get(CATEGORY_COL, "")

        if not harmful_prompt:
            continue

        log(f"\n=== prompt_id={prompt_id} | category={category} ===")
        log(f"    промпт: {harmful_prompt[:120]}...")

        for ctx_type, ctx_form, ctx_len in combinations:
            if ctx_len == 0:
        # baseline: без контекста
                context = ""
                full_prompt = harmful_prompt
            else:
                context = get_context(contexts, ctx_type, ctx_form, ctx_len)
                full_prompt = build_full_prompt(context, harmful_prompt)

            for model in MODELS:
                for rep in range(1, REPEATS + 1):
                    key = make_key(prompt_id, category, model, ctx_type, ctx_form, ctx_len, rep)

                    if key in done_keys:
                        skipped += 1
                        continue

                    t0 = time.time()
                    response = generate_giga(full_prompt, model)
                    dt = time.time() - t0

                    record = {
                        "prompt_id": prompt_id,
                        "model": model,
                        "ctx_type": ctx_type,
                        "ctx_form": ctx_form,
                        "ctx_len": ctx_len,
                        "repeat": rep,
                        "prompt": harmful_prompt,
                        "response": response,
                        "category": category,
                        "latency_sec": round(dt, 3),
                        "timestamp": time.time(),
                    }
                    append_record(CHECKPOINT_FILE, record)
                    done_keys.add(key)
                    total_requests += 1

                    log(f"  ✔ {model} | {ctx_type}/{ctx_form}/{ctx_len} | "
                        f"rep {rep}/{REPEATS} | {dt:.2f}s | "
                        f"len(resp)={len(response)}")

                    time.sleep(SLEEP_BETWEEN_REQUESTS)

    log("\n" + "=" * 70)
    log(f"Готово. Новых запросов: {total_requests}. Пропущено: {skipped}.")


if __name__ == "__main__":
    try:
        main()
    except KeyboardInterrupt:
        log("Прервано пользователем. Прогресс сохранён.")
    except Exception:
        log("Фатальная ошибка:\n" + traceback.format_exc())
```
